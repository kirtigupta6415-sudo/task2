```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
sns.countplot(x="Survived",data=df) #survival count
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x="Sex",hue="Survived",data=df)    # which gender survived the most
plt.title("Gender vs Survived")
plt.show() 
```

```python
sns.countplot(x="Pclass",hue="Survived",data=df)     # which passenger class survived the most
plt.title("Passenger class vs Survived")
plt.show()
```

```python
plt.hist(df["Age"],bins=20)       # age group present in the titanic
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm")   #  which suffer positive and negative survival
plt.title("Correlation heatmaps")
plt.show()
```

```python
sns.boxplot(x="Pclass",y="Fare",data=df)    #  which class cost the most
plt.title("Fare by passenger class")   
plt.show()
```

```python
sns.countplot(x="Embarked",hue="Survived",data=df)    # which embarked people survived the most
plt.title("Embarked vs Survival")
plt.show()
```
